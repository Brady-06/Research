### 一、知识蒸馏（KD）原理**原理**- **知识蒸馏（Knowledge Distillation, KD）**：用一个大的「教师」模型（teacher）指导小的「学生」模型（student）训练，让学生输出分布对齐教师，从而用更小模型逼近教师精度。- 与剪枝的关系：剪枝掉精度后，可用蒸馏把知识从原模型/大模型补回来。**名词**- **软标签（soft label）**：教师输出的概率分布（不是 one-hot 硬标签），含类别之间的相似信息。- **温度（temperature, T）**：softmax 里除以 T，T 越大分布越「软」，能放大教师输出的细节。- **蒸馏损失**：学生输出与教师软标签的 KL 散度（或类似度量），通常和原始检测损失加权求和。
